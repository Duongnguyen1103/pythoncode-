```python
#16
n = int(input())
chan = 0
le = 0
so = n//10
if n%2==0:
    chan += 1
else:
    le += 1
while so>0:
    if so%2==0:
        chan += 1
    else:
        le += 1
    so = so//10
print(f'chan:{chan}\nle:{le}')
```

```python
#17
n = int(input())
a = n//10
sc = n - 10*(n//10)
tong = 0
while a>0:
    tong += sc
    sc = a - 10*(a//10)
    a = a//10
    if a//10==0:
        tong = tong + a
print(tong)
```

```python
#18
h = int(input())
for i in range(1,h+1):
    if i == 1:
        print(" "*(h-1)+"*")
    elif i == h:
        print(" "*(h-i)+"*"*(2*h-1))
    else:
        print(" "*(h-i)+"*"+" "*(2*i-3)+"*")
```
