```python
print("A16")
n = float(input())
if n>0 and n != int(n):
    if n<int(n)+0.5:
        print(f'{int(n)+1:.0f} {int(n):.0f} {int(n):.0f}')
    elif n>=int(n)+0.5:
        print(f'{int(n)+1:.0f} {int(n):.0f} {int(n)+1:.0f}')
elif n<0 and n != int(n):
    if n<int(n)-0.5:
        print(f'{int(n):.0f} {int(n)-1:.0f} {int(n)-1:.0f}')
    elif n>=int(n)-0.5:
        print(f'{int(n):.0f} {int(n)-1:.0f} {int(n):.0f}')
elif n == int(n):
    print(f'{n:.0f} {n:.0f} {n:.0f}')
```
